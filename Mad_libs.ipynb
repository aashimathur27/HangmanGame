```python
print("Welcome to MadLibs! Let's create a fun story together.")
print("You will be asked to provide a few words, and we will use them to generate a silly story.")

name = input("Enter a name: ")
place = input("Enter a place: ")
adjective = input("Enter an adjective: ")
noun1 = input("Enter a noun: ")
verb = input("Enter a verb ending in -ing: ")
animal = input("Enter an animal: ")
noun2 = input("Enter another noun: ")

story = f"""
Once upon a time, there was a person named {name} who lived in {place}. 
One day, {name} woke up feeling very {adjective}. They looked out the window and saw a {noun1}.
Without thinking, {name} started {verb} with excitement! Suddenly, a {animal} appeared, carrying a {noun2}.
It was the beginning of an adventure that {name} would never forget!
"""

print("\nHere's your MadLibs story:")
print(story)
```
